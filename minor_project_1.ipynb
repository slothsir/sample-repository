```python
from datetime import datetime

def streak():
  for name in names:
    dates = []
    for i in messages:
      if i["sender"] == name:
         date = datetime.strptime(i["date"].strip(),'%d/%m/%y,').date()
         if date not in dates:
          dates.append(date)
    dates.sort()
    longest=0
    for i in range(1, len(dates)):
      gap = (dates[i] - dates[i-1]).days - 1
      if gap > longest:
        longest = gap
    print("  ",name,":",longest,"days")

# streAK() for finding the longest streaks and the longest streak goes to vikas
def meowo():
   times = {}
   for i in range(1, len(messages)):
       if messages[i]["sender"] != messages[i-1]["sender"]:
           t1 = datetime.strptime(
               messages[i-1]["date"].strip() + " " + messages[i-1]["timestamp"].strip(),
               "%d/%m/%y, %H:%M"
           )

           t2 = datetime.strptime(
               messages[i]["date"].strip() + " " + messages[i]["timestamp"].strip(),
               "%d/%m/%y, %H:%M"
           )

           gap = (t2 - t1).total_seconds() / 60
           name = messages[i]["sender"]
           if name not in times:
               times[name] = []
           times[name].append(gap)
   for name in times:
       average = sum(times[name]) / len(times[name])
      # print(name, ":", round(average, 1), "minutes")
   fastest = 999999
   fastest_name = ""
   for name in times:
       average = sum(times[name]) / len(times[name])
       if average < fastest:
           fastest = average
           fastest_name = name
   print("  Fastest replier :", fastest_name, "(avg", round(fastest, 1), "minutes)")
   # meow is used to find the fastest reply but complex and perplexing
```
